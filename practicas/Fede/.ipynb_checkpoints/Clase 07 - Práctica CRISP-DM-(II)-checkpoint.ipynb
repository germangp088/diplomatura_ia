**4. Support Vector Machines**

```python
from sklearn.svm import SVC
```

(SVM) son algoritmos de aprendizaje supervisado que se pueden usar para resolver problemas de clasificación, regresión y detección de valores atípicos. Funcionan encontrando un hiperplano que separa los puntos de datos de una clase de los de otra clase, siendo el mejor hiperplano el que tiene el mayor margen entre las dos clases.
Son efectivas en espacios muchas dimensiones, incluso cuando el número de dimensiones es mayor que el número de muestras.
Aplicaciones: procesamiento de señales, aplicaciones médicas, procesamiento del lenguaje natural y reconocimiento de voz e imágenes.
Funcionan correlacionando datos con un espacio de características de grandes dimensiones para que los puntos de datos se puedan categorizar, incluso si los datos no se pueden separar linealmente de otras maneras.
Para entrenarlo, le proporciona conjuntos de datos de entrenamiento etiquetados para cada categoría, lo que le permite categorizar nuevas muestras.
Cuando los datos no son perfectamente separables o contienen valores atípicos, SVM permite una técnica de margen suave que introduce una variable de holgura para cada punto de datos, lo que suaviza el estricto requisito de margen y permite ciertas clasificaciones erróneas.

```python
svm_model = SVC()
svm_model = svm_model.fit(X_train, y_train)
```

```python
y_prediction_svm = svm_model.predict(X_test)
print(classification_report(y_test, y_prediction_svm))
```

**5. Red Neuronal**

```python
import tensorflow as tf
from keras.models import Sequential
from keras.layers import Dense

print(tf.__version__)
```

```python
def plot_loss(history):
  fig, (ax1, ax2) = plt.subplots(1,2)
  ax1.plot(history.history['loss'], label='loss')
  ax1.plot(history.history['val_loss'], label='val_loss')
  ax1.set_xlabel('Epoch')
  ax1.set_ylabel('Binary crossentropy')
  ax1.legend()
  ax1.grid(True)

  ax2.plot(history.history['accuracy'], label='accuracy')
  ax2.plot(history.history['val_accuracy'], label='val_accuracy')
  ax2.set_xlabel('Epoch')
  ax2.set_ylabel('Accuracy')
  ax2.legend()
  ax2.grid(True)
  plt.show()
```

En el proceso de entrenamiento, la función sigmoidea se utiliza en la capa de salida porque este conjunto de datos es un problema de clasificación.

Se utiliza la entropía cruzada binaria ya que la salida del modelo es una probabilidad entre 0 y 1 (en este caso, si el paquete llegó a tiempo o no).

Adam se utiliza como una de las opciones más populares para problemas de clasificación y estamos interesados en la precisión del modelo.

```python
def train_nn_model(X_train, y_train, num_nodes, dropout_rate, lr, batch_size, epochs, X_validation, y_validation):


#model = tf.keras.models.Sequential([
#    tf.keras.layers.InputLayer(input_shape=(1,)),
#    tf.keras.layers.Dense(1)
#])

    nn_model = tf.keras.Sequential([
#        tf.keras.layers.InputLayer(shape=(16,)),
        tf.keras.layers.InputLayer(input_shape=(16,)),
        tf.keras.layers.Dense(num_nodes, activation='relu'),
        tf.keras.layers.Dropout(dropout_rate),
        tf.keras.layers.Dense(num_nodes, activation='relu'),
        tf.keras.layers.Dropout(dropout_rate),
        tf.keras.layers.Dense(1, activation='sigmoid')
    ])

    nn_model.compile(loss='binary_crossentropy', optimizer=tf.keras.optimizers.Adam(lr), metrics=['accuracy'])

    history = nn_model.fit(X_train, y_train, epochs=epochs, batch_size=batch_size, validation_data=(X_validation, y_validation), verbose=0)

    return nn_model, history
```

El mejor modelo tiene 32 nodos,

una tasa de aprendizaje de 0,005

una tasa de abandono de 0,2 y un lote de tamaño de 64.


```python
least_val_loss = float('inf')
least_loss_model = None

epochs=10

#cantidad de nodos
#for num_nodes in [16, 32, 64]:
num_nodes = 32

#tasa de abandono
#    for dropout_rate in[0, 0.2]:
dropout_rate = 0.2
#        for lr in[0.1, 0.005, 0.001]:

#tasa de aprendizaje
lr = 0.005
#          for batch_size in[32, 64, 128]:
batch_size = 64

print(f"{num_nodes} nodes, lr {lr}, dropout rate {dropout_rate}, batch size {batch_size}")
model, history = train_nn_model(X_train=X_train, y_train=y_train, num_nodes=num_nodes, dropout_rate=dropout_rate, lr=lr, batch_size=batch_size, epochs=epochs, X_validation=X_val, y_validation=y_val)
```

```python
plot_loss(history)
val_loss = model.evaluate(X_test, y_test)[0]
if val_loss < least_val_loss:
  least_val_loss = val_loss
  least_val_model = model
```

En el proceso de entrenamiento, uso la función "Sigmoidea" en la capa de salida porque este Dataset sirve para clasificación.
Uso entropía cruzada binaria porque la salida del modelo es una probabilidad entre 0 y 1.
El optimizador Adam(lr) se usa para problemas de clasificación y porque estamos interesados en la precisión del modelo.

```python
history.history['val_accuracy'][-1]
```

```python
history.history['val_loss'][-1]
```

```python
model.evaluate(X_test, y_test)
```

El modelo no se sobreajusta y su precisión en los datos de prueba indica un buen rendimiento en datos invisibles a pesar de la menor precisión en comparación con la precisión de validación.

**6. Random forest classification**


```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score, ConfusionMatrixDisplay
from sklearn.model_selection import RandomizedSearchCV, train_test_split
from scipy.stats import randint

from sklearn.tree import export_graphviz
from IPython.display import Image
import graphviz
```


"Random forest classification" es un algoritmo de aprendizaje supervisado que utiliza múltiples árboles de decisión para clasificar datos.
El algoritmo crea árboles de decisión utilizando diferentes subconjuntos aleatorios de datos y características. Luego, cada árbol de decisión proporciona su opinión sobre cómo clasificar los datos.
El algoritmo realiza predicciones calculando la predicción para cada árbol de decisión y tomando el resultado más "popular".

Dado que "Random forest classification" no necesita normalización, dividiremos el conjunto de datos original antes de la normalización.

```python
X_df = df[df.columns[:-1]].values
y_df = df[df.columns[-1]].values

oversampler = RandomOverSampler(random_state=42)

X_resampled_df, y_resampled_df = oversampler.fit_resample(X_df, y_df)

X_train_df, X_temp_df, y_train_df, y_temp_df = train_test_split(X_resampled_df, y_resampled_df, test_size=0.2, random_state=42)
X_val_df, X_test_df, y_val_df, y_test_df = train_test_split(X_temp_df, y_temp_df, test_size=0.5, random_state=42)
```

```python
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train_df, y_train_df)
```

```python
y_prediction_rf = rf.predict(X_test_df)
print(classification_report(y_test_df, y_prediction_rf))
```

```python
for i in range(3):
    tree = rf.estimators_[i]
    dot_data = export_graphviz(tree,
                               feature_names=df.columns[:-1],
                               filled=True,
                               max_depth=2,
                               impurity=False,
                               proportion=True)
    graph = graphviz.Source(dot_data)
    display(graph)
```

Después de visualizar nuestro árbol, podemos ajustar algunos "hiperparámetros"

n_estimators - número de árboles en el bosque; aumentar esto puede mejorar el rendimiento pero aumentará el tiempo necesario para entrenar y predecir.

(*rand_search.fit(X_train_df, y_train_df)* puede demorar varios segundos)

```python
param = {'n_estimators' : randint(50,500)}

rand_search = RandomizedSearchCV(rf, param_distributions = param, n_iter = 5, cv = 5)

rand_search.fit(X_train_df, y_train_df)

best_rf = rand_search.best_estimator_

print('Best hyperparameters:',  rand_search.best_params_)
```

Los mejores hiperparámetros encontrados por el algoritmo son 463 árboles en un bosque.

El siguiente modelo muestra los mejores parámetros.

```python
best_rf = RandomForestClassifier(n_estimators=350, random_state=42)
#best_rf = RandomForestClassifier(n_estimators=169, random_state=42)
#best_rf = RandomForestClassifier(n_estimators=122, random_state=42)
best_rf.fit(X_train_df, y_train_df)

y_best_prediction_rf = best_rf.predict(X_test_df)
print(classification_report(y_test_df, y_best_prediction_rf))
```

Dado que la cantidad de árboles no mejora el rendimiento, se establecerá en 350 para mejorar el tiempo de ejecución.

```python
cm = confusion_matrix(y_test_df, y_best_prediction_rf)
ConfusionMatrixDisplay(confusion_matrix=cm).plot();
```

Al observar la matriz de cofusión, podemos ver que el modelo predice mejor cuando el paquete no llega a tiempo (1) que cuando el paquete llega a tiempo (0).

Podemos ver que predice 0 veces más que 1, lo que indica que hay sesgo presente.

El sesgo intentará normalizarse utilizando el conjunto de datos original sin sobremuestrear la variable de salida.

```python
feature_importances = pd.Series(best_rf.feature_importances_, index=df.columns[:-1]).sort_values(ascending=False)
feature_importances.plot.bar();
```

Dividiendo el conjunto de datos original, sin sobremuestrear, para eliminar sesgos...

```python
X_no = df[df.columns[:-1]].values
y_no = df[df.columns[-1]].values

X_train_no, X_temp_no, y_train_no, y_temp_no = train_test_split(X_no, y_no, test_size=0.2, random_state=42)
X_val_no, X_test_no, y_val_no, y_test_no = train_test_split(X_temp_no, y_temp_no, test_size=0.5, random_state=42)
```

```python
best_rf_no = RandomForestClassifier(n_estimators=350, random_state=42)
best_rf_no.fit(X_train_no, y_train_no)

y_best_prediction_rf_no = best_rf_no.predict(X_test_no)
print(classification_report(y_test_no, y_best_prediction_rf_no))
```

```python
cm_no = confusion_matrix(y_test_no, y_best_prediction_rf_no)
ConfusionMatrixDisplay(confusion_matrix=cm_no).plot();
```

Al eliminar la parte de sobremuestreo antes de dividir los datos,
el sesgo para predecir 1 se ha reducido ligeramente, a costa de reducir la precisión general del modelo en un 8%.
