## Ejercicio 1
La juguetería Gaby desea estimar mediante regresión lineal simple las ventas para el mes
de Julio de su nuevo carrito infantil "Mate".

La información del comportamiento de las
ventas de todos sus almacenes de cadena se presenta en el siguiente tabulado: ventas.csv

Determinar el pronóstico para el mes de venta nro. 7.

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split 
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df = pd.read_csv("../Clase 09 - ventas.csv", sep=",", encoding="latin-1")
CR= df["Mes"].corr(df['Ventas'])
X= df['Mes'].values.reshape(-1,1)
y= df['Ventas'].values.reshape(-1,1)
regressor = LinearRegression()
regressor.fit(X, y)
y_Pred = regressor.predict(X)
print("\nCoeficiente correlacion:\n\t %.3f" %(CR))
print('\nPendiente:\n\t %0.2f' %regressor.coef_[0][0])
print('\nIntecept:\n\t %0.2f' %regressor.intercept_[0])
print('\nMean squared error:\n\t %.2f' % mean_squared_error(y, y_Pred))
print('\nCoefficient of determination:\n\t %.2f' % r2_score(y, y_Pred))
print("\nPredicción de venta Mes 7 =\n\t %.2f" % regressor.predict([[7]])[0][0])
plt.scatter(X, y,  color='black')
plt.plot(X, y_Pred, color='blue', linewidth=3)
plt.xticks(())
plt.yticks(())
plt.show()
print('y(Ventas) = (%0.2f * X) + %0.2f' % (regressor.coef_[0][0], regressor.intercept_[0]))
```

```python
from sklearn.datasets import load_iris
from sklearn import tre
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import confusion_matrix

iris = load_iris()

X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target,test_size=0.4, random_state=0)

clf = tree.DecisionTreeClassifier()

clf = clf.fit(X_train, y_train)

print("Accuracy1: %0.2f" % clf.score(X_test, y_test))

clf.predict(X_test)

clf = tree.DecisionTreeClassifier()

scores = cross_val_score(clf, iris.data,iris.target, cv=5)

print("Accuracy2: %0.2f (+/- %0.2f)" % (scores.mean(), scores.std() * 2))

y_pred = clf.fit(X_train,y_train).predict(X_test)

conf_matrix_1 = confusion_matrix(y_test, y_pred)

print(conf_matrix_1)

y_pred = cross_val_predict(clf, iris.data,iris.target, cv=5)

conf_matrix_2 = confusion_matrix(iris.target, y_pred)

print(conf_matrix_2)
```
