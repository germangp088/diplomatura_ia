```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from xgboost import XGBClassifier 
```

```python
#Cargo el dataset en una variable
dataset=pd.read_excel("../smoking_prediction.xlsx")
```

Matriz de correlación para identificar señales útiles y redundancias.


# Selección de features y limpieza (EDA)

Para reducir overfitting y evitar ruido, removo variables con baja correlación con *smoking* o redundantes por alta multicolinealidad:

**Drop:**
- `ID` (identificador, sin poder predictivo)
- `eyesight(left)`, `eyesight(right)` y `hearing(left)`, `hearing(right)` (sin relación con fumar en este dataset)
- `oral` (ruido en este dataset)
- `Cholesterol` y `LDL` (redundantes con `HDL` — multicolinealidad)
- `Urine protein`
- `smoking` (target, para evitar data leakage)

Mantengo `tartar` y `gender` como categóricas (se codifican con One‑Hot Encoder).



## Selección final de variables (basado en el heatmap)
Se eliminan por **baja correlación** con *smoking* o **redundancia** entre sí:
- `eyesight(left)`, `eyesight(right)`, `hearing(left)`, `hearing(right)`, `Urine protein`, `hemoglobin`, `AST`, `ALT`, `dental caries`
- `HDL` y `LDL` (alta correlación con otros lípidos). **Se mantiene `Cholesterol`**.
- `ID` y `oral` por no aportar al objetivo.
- `smoking` se elimina de `X` para evitar *data leakage*.

Se mantienen como categóricas: `tartar` y `gender` (se codifican con One‑Hot).
El `test_size` se fija en **0.25** con `stratify=y` para una evaluación más robusta.


```python
cols_to_drop = ['ID', 'oral', 'eyesight(left)', 'eyesight(right)', 'hearing(left)', 'hearing(right)', 'Urine protein', 'hemoglobin', 'AST', 'ALT', 'dental caries', 'HDL', 'LDL', 'smoking']
cols_to_train = ["tartar", "gender"]
```

```python
#Codificación y creación de train y test y dropeo la variable ID ya que no aporta valor predictivo y solo genera ruido en el modelo.
X = dataset.drop(columns=cols_to_drop, axis=1).copy()
y = dataset.smoking.copy()
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.255, random_state=42, stratify=y)
```

```python
#Voy a hacer uan gráfica para poder representar y entender un poco mejor la composición de fumadores y no fumadores
smoker_map = dataset.groupby(['gender', 'smoking']).size().reset_index(name='count')
smoker_map['smoking'] = smoker_map['smoking'].map({0: 'No fuma', 1: 'Fuma'})
sns.set(style="whitegrid")
plt.figure(figsize=(8,6))
sns.barplot(data=smoker_map, x='gender', y='count', hue='smoking', palette='Set2')
plt.title('Comparativa de fumadores por género')
plt.xlabel('Género')
plt.ylabel('Cantidad de personas')
plt.legend(title='¿Fuma?')
plt.tight_layout()
plt.show()
```

```python
X_train.head(5)
```

```python
ohe = OneHotEncoder(sparse_output=False).fit(X_train[cols_to_train])
```

```python
encoded = ohe.transform((X_train[cols_to_train]))
```

```python
encoded_df = pd.DataFrame(columns=ohe.get_feature_names_out(), data = encoded, index=X_train.index)
```

```python
X_train = pd.concat([X_train, encoded_df], axis="columns")
```

```python
X_train = X_train.drop(cols_to_train, axis=1)
```

```python
#Replico el trabajo hecho con las dummies en el test
encoded_test = ohe.transform((X_test[cols_to_train]))
```

```python
encoded_df_test = pd.DataFrame(columns=ohe.get_feature_names_out(), data = encoded_test, index=X_test.index)
```

```python
X_test = pd.concat([X_test, encoded_df_test], axis="columns")
```

```python
X_test = X_test.drop(cols_to_train, axis=1)
```

```python
#Chequeo que ahora todas las variables sean compatibles on los árboles
X_train.info()
```

```python
X_test.info()
```


## Split de entrenamiento y prueba
Se ajusta el `test_size` a **0.25** para una evaluación más robusta manteniendo `stratify=y`.


```python
xgModel = XGBClassifier()  #creo el modelo
```

```python
#Hago las prediciones
train_pred = xgModel.predict(X_train)
test_pred = xgModel.predict(X_test)
```

```python
#Veo las analíticas para comparar los resultados
print("Train")
print(classification_report(y_train, train_pred))
print("----------------------------------------------------------")
print("Test")
print(classification_report(y_test, test_pred))
```

```python
#Con xgboost en la precisión vemos que en el train cuando una persona no fumo el modelo acierta un 91% de las veces, mientras que cuando fuma la persona el modelo acierta un 82%, si lo comparamos en el test los no fumadores acierta un 83% y los fumadores un 68%. Con respecto al recall, vemos que en el train acertó en el 89% del total de los que no fuman y un 84% en los que sí. mientras que en el test acertó en el 81% del total de los que no fuman y un 71% en los que sí. Si bien el accuracy da un %87 en el train y un 77% en el test, preocupa que el modelo falla mucho al acertar si la persona es fumadora.
```

```python
#Los resultados fueron peores que con los parámetros calculados a ojo con elprimer modelo.
#Voy a cargar el DF con lo datos para entrega
results = pd.read_excel("../smoking_prediction_entrega.xlsx")
```

```python
#Trabajo nuevamente el dataset para que pueda trabajar el modelo en las mismas condiciones que el dataset de entrenamiento. Genero las dummies y dropeo las columnas que no se unan. Seleccioné el modelo del Random Forest con las variables relevantes porque fue el modelo que me dió la mejor proporción accuracy en train y test.
encoded_ohe_entrega = ohe.transform(results[["tartar", "gender"]])
encoded_df_entrega = pd.DataFrame(columns=ohe.get_feature_names_out(), data=encoded_ohe_entrega, index=results.index)
# Elimino columnas no usadas por el modelo
X_entrega = results.drop(["ID", "oral", "weight(kg)", "height(cm)", "Urine protein", "dental caries", "hearing(right)", "hearing(left)", "gender", "tartar"], axis=1).copy()

# Uno las dummies
X_entrega_final = pd.concat([X_entrega, encoded_df_entrega], axis=1)

# Reordeno columnas según el modelo y elimino las features no utilizadas
X_entrega_final = X_entrega_final.reindex(columns=xgModel.feature_names_in_, fill_value=0)

# Genero las predicciones
predicciones = xgModel.predict(X_entrega_final)

# Agrego la columna al DataFrame original
results["smoking"] = predicciones
```

```python
#Exporto al excel para la entrega
results.to_excel("dataset_output.xlsx", index=False)
```
