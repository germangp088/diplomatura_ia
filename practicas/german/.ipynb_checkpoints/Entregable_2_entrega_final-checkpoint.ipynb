
# Predicción de fumadores — Informe

**Objetivo.** Predecir si una persona es fumadora (`smoking`) a partir de variables clínicas y demográficas.  
El entregable incluye: carga de datos, limpieza/EDA, selección de variables, entrenamiento/evaluación y predicción final.


```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from xgboost import XGBClassifier 
```

```python
dataset=pd.read_excel("../smoking_prediction.xlsx")
```

```python
dataset.head(5)
```

```python
dataset.info()
```


## Criterio de EDA y selección de variables (basado en heatmap)

El **heatmap de correlación** se utiliza con un alcance acotado para este trabajo:

- **Eliminar features muertos**: columnas sin relación aparente con `smoking` (p.ej., `eyesight(left/right)`, `hearing(left/right)`, `Urine protein`, `dental caries`, `AST`, `ALT`).  
- **Eliminar IDs o columnas sin sentido predictivo** (p.ej., `ID`, `oral`).  
- **Evitar *data leakage***: `smoking` se elimina de `X` antes de entrenar.

> Nota: El heatmap **no** se usa para estimar importancia del modelo. XGBoost captura relaciones **no lineales** y **umbrales**, por lo que variables con baja correlación lineal (p.ej., `Gtp`) pueden resultar útiles para el modelo.



## Matriz de correlación (criterio de selección de variables)

**Objetivo del heatmap**:
- **Eliminar features totalmente muertos** (sin relación aparente con la variable objetivo).
- **Eliminar IDs o columnas sin sentido predictivo**.

**Decisiones basadas en el heatmap y el criterio anterior:**
- Se eliminan `eyesight(left/right)` y `hearing(left/right)`, `Urine protein`, `dental caries`, `AST`, `ALT` por ausencia de relación con *smoking*.
- Se elimina `ID` por no ser un predictor válido.
- Se elimina `smoking` de `X` para evitar **data leakage**.
- Se conservan variables que el modelo (XGBoost) pondera como útiles aunque no tengan correlación lineal fuerte (p.ej., `Gtp`), porque los árboles capturan **relaciones no lineales y umbrales**.



```python

plt.figure(figsize=(12,10))
sns.heatmap(dataset.corr(numeric_only=True), cmap="coolwarm", annot=False)
plt.title("Correlation matrix (numeric features)")
plt.show()
```

```python
dataset.isnull().mean()*100
```

```python
dataset.smoking.value_counts(normalize=True).mul(100)
```

```python
cols_to_drop = ['ID', 'oral', 'eyesight(left)', 'eyesight(right)', 'hearing(left)', 'hearing(right)', 'Urine protein', 'dental caries', 'AST', 'ALT', 'smoking']
cols_to_train = ["tartar", "gender"]
```

```python
X = dataset.drop(columns=cols_to_drop, axis=1).copy()
y = dataset.smoking.copy()
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)
```

```python

tmp = (
    dataset
    .replace({'gender': {'F':'Femenino','M':'Masculino'}, 'smoking': {0:'No fuma', 1:'Fuma'}})
    .groupby(['gender','smoking']).size()
    .reset_index(name='count')
)

order_gender = ['Femenino','Masculino']
order_smoke = ['No fuma','Fuma']

plt.figure(figsize=(8,5))
sns.set_style("whitegrid")
ax = sns.barplot(data=tmp, x='gender', y='count', hue='smoking',
                 order=order_gender, hue_order=order_smoke)

ax.set_title('Distribución de fumadores por sexo', fontsize=14, pad=12)
ax.set_xlabel('Sexo', fontsize=12)
ax.set_ylabel('Cantidad de personas', fontsize=12)
ax.legend(title='¿Fuma?', frameon=True)

for p in ax.patches:
    height = p.get_height()
    ax.annotate(f'{int(height):,}'.replace(',', '.'),
                (p.get_x() + p.get_width()/2., height),
                ha='center', va='bottom', fontsize=10, xytext=(0,3), textcoords='offset points')

plt.tight_layout()
plt.show()
```

```python
X_train.head(5)
```

```python
ohe = OneHotEncoder(sparse_output=False).fit(X_train[cols_to_train])
```

```python
encoded = ohe.transform((X_train[cols_to_train]))
```

```python
encoded_df = pd.DataFrame(columns=ohe.get_feature_names_out(), data = encoded, index=X_train.index)
```

```python
X_train = pd.concat([X_train, encoded_df], axis="columns")
```

```python
X_train = X_train.drop(cols_to_train, axis=1)
```

```python
encoded_test = ohe.transform((X_test[cols_to_train]))
```

```python
encoded_df_test = pd.DataFrame(columns=ohe.get_feature_names_out(), data = encoded_test, index=X_test.index)
```

```python
X_test = pd.concat([X_test, encoded_df_test], axis="columns")
```

```python
X_test = X_test.drop(cols_to_train, axis=1)
```

```python
X_train.info()
```

```python
X_test.info()
```


## Entrenamiento del modelo


```python
xgModel = XGBClassifier()  #creo el modelo
```

```python
xgModel.fit(X_train, y_train) #entreno el modelo
```

```python
train_pred = xgModel.predict(X_train)
test_pred = xgModel.predict(X_test)
```

```python
print("Train")
print(classification_report(y_train, train_pred))
print("----------------------------------------------------------")
print("Test")
print(classification_report(y_test, test_pred))
```


## Métricas y validación

Se separa el dataset con `test_size=0.25` y `stratify=y`.  
Resultados obtenidos (ejemplo de la última corrida):

- **Train**: accuracy ≈ 0.85, f1 (fumador) ≈ 0.80  
- **Test**: accuracy ≈ 0.76, f1 (fumador) ≈ 0.68

La diferencia entre train y test (~9 pts) es **razonable** para boosting sin *tuning* y sugiere **buena generalización** sin *leakage*.



## Conclusiones

- El modelo XGBoost entrega desempeño **sólido y consistente** en test (accuracy ~0.76, f1 fumador ~0.68).  
- El uso del heatmap se restringió a: eliminar columnas sin valor aparente, quitar IDs y evitar *leakage*.  
- Variables como `Gtp` pueden aportar al modelo aun con correlación lineal baja, debido a **relaciones no lineales**.
- El pipeline de inferencia replica exactamente las transformaciones de entrenamiento (drop + OHE + reindex), y la columna `smoking` se agrega como predicción final para la entrega.


```python
results = pd.read_excel("../smoking_prediction_entrega.xlsx")
```

```python
encoded_ohe_entrega = ohe.transform(results[["tartar", "gender"]])
encoded_df_entrega = pd.DataFrame(columns=ohe.get_feature_names_out(), data=encoded_ohe_entrega, index=results.index)

drop_infer = ['ID', 'oral', 'eyesight(left)', 'eyesight(right)', 'hearing(left)', 'hearing(right)', 'Urine protein', 'dental caries', 'AST', 'ALT', 'smoking', 'gender', 'tartar']
X_entrega = results.drop(columns=[c for c in drop_infer if c in results.columns]).copy()

X_entrega_final = pd.concat([X_entrega, encoded_df_entrega], axis=1)

X_entrega_final = X_entrega_final.reindex(columns=xgModel.feature_names_in_, fill_value=0)

predicciones = xgModel.predict(X_entrega_final)

results["smoking"] = predicciones
```

```python
results.to_excel("dataset_output.xlsx", index=False)
```
