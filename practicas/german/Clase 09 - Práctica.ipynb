Realizar un clasificador Random Forest para predecir si una flor es vertiginosa, setosa o virginica. Utilizar el dataset “iris.csv”.

```python
import requests
import json


BASE_URL = "https://integration-api-staging.seedtalent.com"
LOGIN_ENDPOINT = f"{BASE_URL}/api/v1/auth/login"
GROUPS_ENDPOINT = f"{BASE_URL}/api/v1/groups/"

login_data = {
    "integratorId": "d99a0161-c7ab-462a-9dfa-21dbd1b2853a",
    "apiKey": "3bf85747-9a28-4479-964f-01d9bd3f35fa"
}

def main():
    print("Login in...")
    login_response = requests.post(
        LOGIN_ENDPOINT,
        json=login_data,
        headers={"Content-Type": "application/json"}
    )

    if login_response.status_code != 200:
        print(f"Error en login: {login_response.status_code}")
        print(f"Respuesta: {login_response.text}")
        return
    
    login_result = login_response.json()
    bearer_token = login_result.get("token")
    
    if not bearer_token:
        print("Error: No se recibió token en la respuesta")
        print(f"Respuesta completa: {json.dumps(login_result, indent=2)}")
        return
    
    print(f"Login successful! Token obtained: {bearer_token[:20]}...")
    
    print("\nGetting groups...")
    groups_response = requests.get(
        GROUPS_ENDPOINT,
        headers={
            "Authorization": f"Bearer {bearer_token}",
            "Content-Type": "application/json"
        }
    )
    
    if groups_response.status_code != 200:
        print(f"Error getting groups: {groups_response.status_code}")
        print(f"Response: {groups_response.text}")
        return
    
    groups = groups_response.json()
    print(f"\nGroups obtained successfully!")
    print(f"Total groups: {len(groups)}")
    print(f"\nGroups:")
    print(json.dumps(groups, indent=2, ensure_ascii=False))

if __name__ == "__main__":
    main()

```

# EJERCICIO 3

Dado el siguiente código en python, explique con sus palabras lo que se hace en cada línea de código:

```python
from sklearn.datasets import load_iris
from sklearn import tre
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import confusion_matrix

iris = load_iris()

X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target,test_size=0.4, random_state=0)

clf = tree.DecisionTreeClassifier()

clf = clf.fit(X_train, y_train)

print("Accuracy1: %0.2f" % clf.score(X_test, y_test))

clf.predict(X_test)

clf = tree.DecisionTreeClassifier()

scores = cross_val_score(clf, iris.data,iris.target, cv=5)

print("Accuracy2: %0.2f (+/- %0.2f)" % (scores.mean(), scores.std() * 2))

y_pred = clf.fit(X_train,y_train).predict(X_test)

conf_matrix_1 = confusion_matrix(y_test, y_pred)

print(conf_matrix_1)

y_pred = cross_val_predict(clf, iris.data,iris.target, cv=5)

conf_matrix_2 = confusion_matrix(iris.target, y_pred)

print(conf_matrix_2)
```
